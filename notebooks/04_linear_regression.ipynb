```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/train.csv")
df["LogSalePrice"] = np.log(df["SalePrice"])
```

## Stage 4 — Linear Regression Analysis

In this stage, linear regression is used to model the relationship between house prices and selected numerical features.

The goal is not prediction, but understanding:
- how variables are related to house prices
- how to interpret regression coefficients
- how regression connects to correlation results from the previous stage

### Simple Linear Regression: Price and Overall Quality

Research question:  
How does the overall quality of a house relate to its sale price?

```python
!pip install statsmodels
```

```python
import statsmodels.api as sm

X = df["OverallQual"]
y = df["LogSalePrice"]

X = sm.add_constant(X)  

model = sm.OLS(y, X).fit()
model.summary()
```

### Interpretation of Regression Results

A simple linear regression was fitted using `OverallQual` to explain `LogSalePrice`.

The regression coefficient for `OverallQual` is positive and statistically significant.
This means that houses with higher overall quality tend to have higher sale prices.

The coefficient value (about 0.24) indicates that when the overall quality increases by one unit,
the log-transformed sale price increases by about 0.24.
Because the target variable is in logarithmic form, this can be interpreted as an approximate percentage increase in price.

The R-squared value is about 0.67.
This means that around 67% of the variation in house prices is explained by overall quality alone.

This result shows a strong relationship between house quality and sale price.
However, this model includes only one variable and does not account for other factors such as size, location, or age of the house.

### Multiple Linear Regression: Overall Quality and Living Area

After fitting a simple linear regression model, the analysis is extended to a multiple linear regression model.

In this model, two explanatory variables are used:
- `OverallQual`, which represents the overall quality of the house
- `GrLivArea`, which represents the above-ground living area

The goal of this step is to examine how these variables jointly relate to house prices.
Including more than one variable allows us to see the effect of each variable while holding the other one constant.

The dependent variable remains `LogSalePrice`, which helps reduce skewness and makes the interpretation more stable.

```python
import statsmodels.api as sm

X = df[["OverallQual", "GrLivArea"]]
y = df["LogSalePrice"]

X = sm.add_constant(X)

model_multi = sm.OLS(y, X).fit()
model_multi.summary()
```

### Interpretation of Multiple Regression Results

A multiple linear regression model was fitted using `OverallQual` and `GrLivArea` to explain `LogSalePrice`.

Both variables have positive and statistically significant coefficients.
This means that house quality and living area are both related to higher house prices.

The coefficient for `OverallQual` is smaller than in the simple regression model.
This shows that part of the effect of quality is shared with house size.
However, even after accounting for living area, overall quality remains an important factor in explaining house prices.

The coefficient for `GrLivArea` is also positive.
This indicates that, for houses with the same quality, larger living area is associated with higher prices.

The R-squared value is about 0.74.
This means that approximately 74% of the variation in house prices is explained by this model.
This is an improvement compared to the simple regression model.

Overall, the multiple regression model provides a better explanation of house prices than a single-variable model.
